## Разберемся с Yahoo Finance

Пример URL:
https://query1.finance.yahoo.com/v8/finance/chart/MTF=F?period1=1356998400&period2=1785000000&interval=1d

Запрос к Yahoo Finance:

    Сервер: query1.finance.yahoo.com
    Путь: /v8/finance/chart/
    Тикер: MTF=F (вставляется прямо в путь)

    Параметры запроса:
        period1 — начало периода (Unix-время)
        period2 — конец периода (Unix-время)
        interval — шаг: 1d (день), 1wk (неделя), 1mo (месяц)

**Что такое тикен?**
Тикер (ticker) — это краткое буквенное обозначение актива на бирже (акции, валюты, фьючерса и т.д.).

##### Суффиксы тикеров на Yahoo Finance

**=F** — фьючерс  
Пример: `MTF=F` (уголь), `BZ=F` (Brent), `GC=F` (золото)

**=X** — валютная пара  
Пример: `EURPLN=X`, `USDPLN=X`

**^** в начале — индекс  
Пример: `^GSPC` (S&P 500), `^DJI` (Dow Jones)

**.WA** — Варшавская биржа (Польша)  
Пример: `JSW.WA`, `PKN.WA`

**.DE** — XETRA (Германия)  
Пример: `BMW.DE`, `SAP.DE`

**.L** — Лондонская биржа  
Пример: `VOD.L`, `HSBA.L`

**Без суффикса** — американские биржи (NYSE, NASDAQ)  
Пример: `AAPL`, `MSFT`, `TSLA`

```python
import requests

url = "https://query1.finance.yahoo.com/v1/finance/search?q=coal&quotesCount=10"
data = requests.get(url, headers={"User-Agent": "Mozilla/5.0"}).json()

for item in data["quotes"]:
    print(item["symbol"], "|", item.get("shortname"), "|", item["exchange"])
```
