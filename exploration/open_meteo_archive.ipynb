Как устроен адрес?
Адрес делится знаком "?"

<p>https://archive-api.open-meteo.com/v1/archive  ?  latitude=51.67&longitude=39.21&...</p>
<p>        └────────── куда идём ──────────┘         └──── что именно просим ────┘</p>

После "?" идут парамерты форматом "имя=значение", которые склеены знаком "&"
Порядок не важен, пробелов быть не должно

Параметр	Значение
```
latitude=51.67
longitude=39.21
start_date=2020-06-01
end_date=2020-06-05
```

### Начнем читать этот json

````python
import requests
import json

url = "https://archive-api.open-meteo.com/v1/archive?latitude=54.735152&longitude=55.958736&start_date=2025-12-01&end_date=2026-01-01&daily=sunshine_duration,daylight_duration,shortwave_radiation_sum,cloud_cover_mean,precipitation_sum,rain_sum,snowfall_sum,precipitation_hours&timezone=Asia/Yekaterinburg"
response = requests.get(url)

print("числовой код ответа сервера", response.status_code)
data = json.loads(response.text)

print(json.dumps(data, indent=2))
````

````python
# На последок выясним какая была минимальная температура c 1940 года

import json
import requests

url = "https://archive-api.open-meteo.com/v1/archive?latitude=54.735152&longitude=55.958736&start_date=1940-01-01&end_date=2026-08-18&daily=temperature_2m_min&timezone=Asia/Yekaterinburg"
response = requests.get(url)
data = json.loads(response.text)

print(response.status_code)
print("Кол-во дней: ", len(data["daily"]["time"]))
print("Перечислим ", (data["daily"].keys()))

# мин температура
print("Мин t: ", min(data["daily"]["temperature_2m_min"]))

# Я думал температура ниже, вытащим его дату
# я тут подумал а что если несколько дней были с такой температурой

# посмотрим индексы и вытащим даты

min_temp = min(data["daily"]["temperature_2m_min"])

indices = [i for i, v in enumerate(data["daily"]["temperature_2m_min"]) if v == min_temp]

print(indices)
for i in indices:
    print("Даты: ", data["daily"]["time"][i])
````
